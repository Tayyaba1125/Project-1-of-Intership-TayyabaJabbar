```python
import numpy as np
import pandas as pd
```

```python
dataset=pd.read_excel("Dataset for Data Analytics.xlsx")
df1=pd.DataFrame(dataset)
df1.head()
```

```python
df1.describe()
```

```python
df1.shape
```

```python
#identifying null or missing valued
df1.isnull().sum()
```

```python
#filling the null values in couponcode with no coupon
df1['CouponCode']=df1['CouponCode'].fillna('No Coupon')
```

```python
#verifying the coupncode column
df1['CouponCode'].isnull().sum()
```

```python
#identifying duplicate rows
df1.duplicated().sum()
```

```python
#checking that the values in orderid are unique
df1['OrderID'].is_unique
```

```python
#Proper case and Trim Whitespace
text_cols = ['Product', 'PaymentMethod', 'OrderStatus', 'ReferralSource', 'ShippingAddress']
for c in text_cols:
    df1[c] = df1[c].str.strip()
for c in ['Product', 'PaymentMethod', 'OrderStatus', 'ReferralSource']:
    df1[c] = df1[c].str.title()
```

```python
#rounding the numbers
df1['UnitPrice'] = df1['UnitPrice'].round(2)
df1['TotalPrice'] = df1['TotalPrice'].round(2)
```

```python
#verifying the unitprice
df1['UnitPrice'].head()
```

```python
#verifying the Total price
df1['TotalPrice'].head()
```

```python
#Changing the Date format to (YYYY-MM-DD)
df1['Date'] = pd.to_datetime(df1['Date'], errors='coerce')
print("Invalid dates:", df1['Date'].isnull().sum())
df1['Date'] = df1['Date'].dt.strftime('%Y-%m-%d')
```

```python
#verfying the date format
df1['Date'].head()
```

```python
#Final Verification
print("Duplicate rows:", df1.duplicated().sum())
print("Duplicate OrderIDs:", df1['OrderID'].duplicated().sum())
print("Bad date format:", (~df1['Date'].str.match(r'^\d{4}-\d{2}-\d{2}$')).sum())
print("Remaining nulls:\n", df1.isnull().sum())
```

```python
#Saving the cleaned file
df1.to_excel("Cleaned_Dataset.xlsx", index=False)
```
